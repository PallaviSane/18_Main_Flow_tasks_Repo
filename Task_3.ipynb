```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("householdtask3.csv")
```

```python
data.head()
```

```python
#scatter Chart
plt.scatter(data['year'],data['own'])

#for adding title
plt.title("Scatter Chart")

#for adding labels
plt.xlabel('Year')
plt.ylabel('Own')

#for showing result
plt.show()
```

```python
#line chart of year against own
plt.plot(data['year'])
plt.plot(data['own'])

#for adding title
plt.title("Line Chart")

#for adding labels
plt.xlabel("Year")
plt.ylabel("Own")

#for showing result
plt.show()
```

```python
#Bar Chart
plt.bar(data['year'],data['own'])

#for adding title
plt.title("Bar Chart")

#for adding labels
plt.xlabel('Year')
plt.ylabel('Own')

#for showing result
plt.show()
```
